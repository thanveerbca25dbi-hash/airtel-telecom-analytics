```python
import sqlite3
import pandas as pd

con = sqlite3.connect(r'C:\Users\thanv\OneDrive\Documents\airtel-telecom-analytics\airtel-telecom-analytics\data\raw\airtel.db') 

def q(sql):
    return pd.read_sql(sql, con)


```

```python
from pathlib import Path

sql_file = Path("../sql/practice/06_customer_features.sql")

with open(sql_file, "r", encoding="utf-8") as f:
    sql_script = f.read()

con.executescript(sql_script)
con.commit()

print("Customer features SQL executed successfully!")
```
